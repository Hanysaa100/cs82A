```python
def multiply_for(a, b):
    result = 0

    for i in range(b):
        result = result + a

    return result
```

```python
multiply_for(7, 6)
```

```python
def multiply_while(a, b):
    result = 0
    count = 0

    while count < b:
        result = result + a
        count = count + 1

    return result
```

```python
multiply_while(7, 6)
```

```python
print("For loop:", multiply_for(7, 6))
print("While loop:", multiply_while(7, 6))
print("Real product:", 7 * 6)
```

Mltiply by adding one number to itself repeatedly inside a loop.once with a for loop and once with a while loop, both agree with the real product. which is 7*6=42
